# ML 2 Lab Program: K-Means Clustering Using the Elbow Method

**Aim:** To implement K-Means clustering on a dataset containing two features — **Study Hours** and **IQ** — and find a suitable number of clusters using the **Elbow Method**.

**Dataset:** 240 sample data points generated for this lab exercise.

**Libraries:** NumPy, Pandas, Matplotlib, and Scikit-learn.

### Steps
1. Create a simple dataset with Study Hours and IQ.
2. Visualise the original data.
3. Calculate WCSS for different values of K.
4. Plot the Elbow Method graph and inspect the WCSS values.
5. Apply K-Means using the selected K.
6. Visualise the final clusters and their centroids.

```python
# Import the required libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans

# Make the results reproducible
np.random.seed(42)
```

## 1. Create the dataset

We generate 240 data points with two features:
- **Study Hours:** Hours studied per day
- **IQ:** IQ score

The generated data is used only for demonstrating K-Means clustering.

```python
# Generate 240 data points with 2 features
X, _ = make_blobs(
    n_samples=240,
    centers=4,
    n_features=2,
    cluster_std=0.9,
    random_state=42
)

# Convert the generated values into realistic-looking ranges
study_hours = np.clip(X[:, 0] * 1.15 + 5, 0, 12)
iq = np.clip(X[:, 1] * 7 + 100, 70, 145)

# Create a DataFrame
df = pd.DataFrame({
    "Study Hours": study_hours,
    "IQ": iq
})

print("First 5 rows of the dataset:")
display(df.head())

print("\\nTotal number of data points:", len(df))
print("\\nDataset information:")
print(df.describe())
```

## 2. Visualise the original dataset

The scatter plot shows all 240 students using Study Hours and IQ.

```python
plt.figure(figsize=(8, 5))
plt.scatter(df["Study Hours"], df["IQ"], s=35)
plt.title("Dataset: Study Hours vs IQ")
plt.xlabel("Study Hours per Day")
plt.ylabel("IQ")
plt.grid(True, alpha=0.3)
plt.show()
```

## 3. Find the optimal K using the Elbow Method

**WCSS (Within-Cluster Sum of Squares)** measures how close the data points are to the centroid of their assigned cluster. Lower WCSS means points are closer to their centroids.

We calculate WCSS for K values from 1 to 10. The suitable K is estimated by looking for the **elbow point**, where the decrease in WCSS starts becoming less significant.

```python
# Use both features for clustering
features = df[["Study Hours", "IQ"]]

# Calculate WCSS for K = 1 to 10
wcss = []

for k in range(1, 11):
    kmeans = KMeans(n_clusters=k, init="k-means++", random_state=42, n_init=10)
    kmeans.fit(features)
    wcss.append(kmeans.inertia_)

# Display the WCSS values clearly
wcss_table = pd.DataFrame({
    "Number of Clusters (K)": range(1, 11),
    "WCSS": wcss
})

display(wcss_table.round(2))
```

```python
# Plot the Elbow Method graph
plt.figure(figsize=(8, 5))
plt.plot(range(1, 11), wcss, marker="o", linestyle="-")
plt.title("Elbow Method")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("WCSS (Within-Cluster Sum of Squares)")
plt.xticks(range(1, 11))
plt.grid(True, alpha=0.3)

# Add WCSS values as labels on the graph
for k, value in zip(range(1, 11), wcss):
    plt.annotate(f"{value:.0f}", (k, value), textcoords="offset points",
                 xytext=(0, 8), ha="center", fontsize=8)

plt.show()
```

**Observation:** The WCSS decreases as K increases. Choose the K value near the point where the curve begins to flatten (the elbow). For this generated dataset, the elbow is expected around **K = 4**. The exact WCSS values are printed in the table above.

## 4. Apply K-Means clustering

Based on the Elbow Method, we use **K = 4** for this demonstration. If your plotted elbow is clearer at another value, update `optimal_k` accordingly.

```python
# Select the number of clusters from the Elbow Method
optimal_k = 4

# Train the K-Means model
kmeans = KMeans(n_clusters=optimal_k, init="k-means++", random_state=42, n_init=10)
df["Cluster"] = kmeans.fit_predict(features)

# Display the clustered data
display(df.head(10))

print("Final WCSS for K =", optimal_k, ":", round(kmeans.inertia_, 2))
```

## 5. Visualise the final clusters

Each colour represents a cluster. The large X markers indicate the cluster centroids.

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df["Study Hours"],
    df["IQ"],
    c=df["Cluster"],
    cmap="viridis",
    s=40
)

# Plot cluster centroids
centroids = kmeans.cluster_centers_
plt.scatter(
    centroids[:, 0],
    centroids[:, 1],
    marker="X",
    s=250,
    c="red",
    edgecolors="black",
    label="Centroids"
)

plt.title(f"K-Means Clustering (K = {optimal_k})")
plt.xlabel("Study Hours per Day")
plt.ylabel("IQ")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

## 6. Result

K-Means clustering was successfully implemented using the Elbow Method on a dataset containing **240 data points** and two features: **Study Hours** and **IQ**.

The WCSS values were calculated for K = 1 to 10 and visualised in the Elbow Method graph. Based on the elbow, K = 4 was selected for the demonstration, and the final clusters were visualised with their centroids.

**Conclusion:** The Elbow Method helps estimate a suitable number of clusters by observing where the WCSS curve starts to flatten.
